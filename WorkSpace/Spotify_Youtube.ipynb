## Dataset Overview
A combined dataset of **Spotify tracks** and their corresponding **YouTube video data**, focusing on the top 10 songs per artist. Includes audio features and engagement metrics.

Key columns:
- *Audio Features:* Danceability, Energy, Speechiness, Acousticness, etc.
- *Performance Metrics:* Spotify Streams; YouTube Views, Likes, Comments
- *Metadata:* Track, Artist, Album, Album Type, Official Video flag, etc.
- Source: [Kaggle – Spotify and YouTube dataset].

## What We can analyze?
- Compare song popularity: streams vs views vs likes
- Audio characteristics vs popularity (e.g. danceability impact)
- Identify trends by artist, album type, or platform


# 📊 Dataset Columns Describtion

## 🎵 Spotify Columns
- **Track**: اسم الأغنية.  
- **Artist**: اسم الفنان أو الفرقة.  
- **Album**: اسم الألبوم الذي تنتمي إليه الأغنية.   
- **Duration_ms**: مدة الأغنية بالثواني أو بالدقائق.  
- **Popularity**: مقياس من 0 لـ 100 يعبر عن مدى شهرة الأغنية على Spotify.  
- **Danceability**: مقياس (0-1) يعبر عن مدى قابلية الأغنية للرقص.  
- **Energy**: مقياس (0-1) يعبر عن قوة وحيوية الأغنية.  
- **Loudness**: درجة ارتفاع الصوت (بوحدة dB).  
- **Speechiness**: مقياس (0-1) يوضح مدى احتواء الأغنية على كلام أو حوار.  
- **Acousticness**: مقياس (0-1) يوضح مدى الطابع الصوتي أو الآكوستيك للأغنية.  
- **Instrumentalness**: مقياس (0-1) يوضح مدى قلة أو انعدام الكلمات (موسيقى فقط).  
- **Liveness**: مقياس (0-1) يوضح مدى احتمال أن تكون الأغنية قد سُجلت في عرض حي.  
- **Valence**: مقياس (0-1) يعبر عن المزاج العام للأغنية (سعيد أو حزين).  
- **Tempo**: سرعة الإيقاع (BPM).
- **Stream**: عدد مرات تشغيل الاغنيه علي Spotify

## 📹 YouTube Columns
- **Views**: عدد المشاهدات على YouTube.  
- **Likes**: عدد الإعجابات على الفيديو.  
- **Comments**: عدد التعليقات على الفيديو.  
- **Licensed**: يوضح ما إذا كانت الأغنية مرخّصة أم لا (True/False).  
- **official_video**: يوضح ما إذا كان الفيديو هو النسخة الرسمية للأغنية (True/False).

### Feature Engineering

```python
#Clustering using PCA and KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Select only the numeric columns
num_cols = df.select_dtypes(include=['int64', 'float64']).columns

# Standardize the features
scaler = StandardScaler()
df_completed_scaled = scaler.fit_transform(df[['Spotify_stream','Views','Likes']])

# Apply PCA
pca = PCA(n_components=2)
principalComponents = pca.fit_transform(df_completed_scaled)

# Determine the number of clusters using the elbow method
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++', max_iter=300, n_init=10, random_state=0)
    kmeans.fit(principalComponents)
    wcss.append(kmeans.inertia_)

plt.plot(range(1, 11), wcss)
plt.title('Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.show()
```

```python
# From the previous elbow chart we can see that a good number of clusters is 6
kmeans = KMeans(n_clusters=6, init='k-means++', max_iter=300, n_init=10, random_state=0)
df['popularity_cluster'] = kmeans.fit_predict(principalComponents)
```

```python
#Check the clusters distribution by Spotify stream mean
df['Spotify_stream'].groupby(df['popularity_cluster']).mean()/ 1e6 # in millions
```

```python
#Plot the clusters distribution by Spotify stream mean
df.groupby('popularity_cluster')['Spotify_stream'].mean().sort_values().plot(
    kind='bar'
    , figsize=(10,5)
    , title='Average Spotify Streams by Popularity Cluster'
   )
plt.show()
```

```python
# Create a dictionary with  the cluster names for the legend
name_mapper = {5: 'SuperPopular', 1: 'High-popularity', 2: 'High-popularity', 0:'Medium-popularity', 4:'Medium-popularity', 3:'Low-popularity'}
# Use the map to create a new column with the new group names
df['popularity_group'] = df['popularity_cluster'].map(name_mapper)
```

### Model Selection and Evaluation

```python
nDF.groupby('popularity_group').count()
```

```python
# Import libraries for modeling
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from xgboost import XGBClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import RandomizedSearchCV
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
from sklearn.model_selection import train_test_split
from imblearn.over_sampling import SMOTE
```

```python
# Preprocessing
encoder = LabelEncoder()
y = encoder.fit_transform(nDF['popularity_group'])
```

```python
# Divide the data into training and test sets with  SMOTE - k_neighbors=4 is set to include super popular as this just have 5 songs
smote = SMOTE(k_neighbors=4)
X_sm, y_sm = smote.fit_resample(X_scaled, y)

X_train_sm, X_test_sm, y_train_sm, y_test_sm = train_test_split(X_sm, y_sm, test_size=0.3, random_state=42)
```

```python
models={
    'LogisticRegression': LogisticRegression(multi_class='multinomial', solver='lbfgs'),
    'SVC': SVC(decision_function_shape='ovo'),
    'XGBClassifier': XGBClassifier()
}
```

```python
accuracy_scores = {}
classification_reports = {}
confusion_matrices = {}

for model_name, model in models.items():
    model.fit(X_train_sm, y_train_sm)
    y_pred = model.predict(X_test_sm)

    # Accuracy
    accuracy = accuracy_score(y_test_sm, y_pred)
    accuracy_scores[model_name] = accuracy

    # Classification report as dictionary
    report = classification_report(y_test_sm, y_pred, output_dict=True)
    classification_reports[model_name] = report

    # Confusion matrix
    cm = confusion_matrix(y_test_sm, y_pred)
    confusion_matrices[model_name] = cm
```

```python
accuracy_df = pd.DataFrame(list(accuracy_scores.items()), columns=['Model', 'Accuracy'])
accuracy_df = accuracy_df.sort_values(by='Accuracy', ascending=False)
accuracy_df
```

```python
best_model = accuracy_df.loc[accuracy_df["Accuracy"].idxmax()]
best_model
```

```python
data = []
for model_name, report in classification_reports.items():
    for class_label, metrics in report.items():
        if isinstance(metrics, dict): # Exclude 'accuracy', 'macro avg', 'weighted avg'
            row = {'Model': model_name, 'Class': class_label}
            row.update(metrics)
            data.append(row)
        else: # Include 'accuracy', 'macro avg', 'weighted avg'
            row = {'Model': model_name, 'Class': class_label, 'value': metrics}
            data.append(row)

metrics_df = pd.DataFrame(data)
```

```python
# Let's create a simpler DataFrame with overall metrics
overall_metrics = []
for model_name, report in classification_reports.items():
    overall_metrics.append({
        'Model': model_name,
        'Accuracy': report['accuracy'],
        'Macro Avg Precision': report['macro avg']['precision'],
        'Macro Avg Recall': report['macro avg']['recall'],
        'Macro Avg F1-score': report['macro avg']['f1-score'],
        'Weighted Avg Precision': report['weighted avg']['precision'],
        'Weighted Avg Recall': report['weighted avg']['recall'],
        'Weighted Avg F1-score': report['weighted avg']['f1-score']
    })

metrics_df = pd.DataFrame(overall_metrics)
print("\nOverall Classification Metrics:")
metrics_df
```

```python
# display the confusion matrix visualizaions and make them in 3 columns and one row
import matplotlib.pyplot as plt
import seaborn as sns
fig, axes = plt.subplots(1, 3, figsize=(14, 3))

# Flatten the axes array for easy indexing
axes = axes.flatten()
# dray now
for i, (model_name, cm) in enumerate(confusion_matrices.items()):
    ax = axes[i]
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax)
    ax.set_title(f'Confusion Matrix - {model_name}')
    ax.set_xlabel('Predicted')
```
